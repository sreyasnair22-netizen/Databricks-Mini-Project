##Ingest Fact data into bronze layer

```python
from pyspark.sql.types import StructType, StructField, StringType, IntegerType, DateType, BooleanType
import pyspark.sql.functions as F
```

```python
catalog_name = 'ecommerce'
bronze_fact = spark.read.format('csv').load('/Volumes/ecommerce/source_data/raw/order_items/landing/*.csv', header=True, inferSchema=True)
display(bronze_fact)
```

```python
bronze_fact = bronze_fact.withColumn('_source_file_name', F.col('_metadata.file_path'))
bronze_fact = bronze_fact.withColumn('_ingested_at', F.current_timestamp())
display(bronze_fact)
```

```python
bronze_fact.write.format('delta').mode('overwrite').option('mergeSchema', 'true').saveAsTable(f'{catalog_name}.bronze.brz_order_items')
```
