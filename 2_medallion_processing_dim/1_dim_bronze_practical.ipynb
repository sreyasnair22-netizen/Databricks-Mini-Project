## Ingest Dimension data to Bronze tables

```python
from pyspark.sql.types import StructType, StructField, StringType, IntegerType, DoubleType, ArrayType, DateType
import pyspark.sql.functions as F
```

```python
catalog_name = 'ecommerce'
```

##Brands

```python
##loading raw data to dataframe
volume_location = '/Volumes/ecommerce/source_data/raw/brands/*.csv'

df = spark.read.format('csv').load(volume_location, header=True, inferSchema=True, delimiter=',')
df = df.withColumn("_source_file", F.col("_metadata.file_path")) \
    .withColumn("_ingested_at", F.current_timestamp())
df.show(5, truncate=False)
```

```python
df.write.format("delta") \
    .mode("overwrite") \
    .option("mergeSchema", "true") \
    .saveAsTable(f"{catalog_name}.bronze.brz_brands")
```

##Date

```python
date_df = spark.read.format('csv').load('/Volumes/ecommerce/source_data/raw/date/*.csv', header=True, inferSchema=True,)
date_df = date_df.withColumn('_source_file_name', F.col('_metadata.file_path')).withColumn('_ingested_at', F.current_timestamp())
date_df.show(10, truncate=False)
```
